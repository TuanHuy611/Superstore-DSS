```python
# Kiểm tra thông tin dataset
df.info()

# Kiểm tra kiểu dữ liệu
print(df.dtypes)

# Kiểm tra giá trị NULL
print("\nMissing values:")
print(df.isnull().sum())

# Kiểm tra dữ liệu trùng lặp
print("\nDuplicate rows:", df.duplicated().sum())

# Thống kê dữ liệu số
display(df.describe())
```

```python
print("Number of unique orders:",
      df['Order ID'].nunique())

print("Number of customers:",
      df['Customer ID'].nunique())

print("Number of products:",
      df['Product ID'].nunique())

print("\nCategories:")
print(df['Category'].unique())

print("\nRegions:")
print(df['Region'].unique())

print("\nShip Modes:")
print(df['Ship Mode'].unique())
```

```python
# Kiểm tra ngày tháng
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    format='%m/%d/%Y',
    errors='coerce'
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'],
    format='%m/%d/%Y',
    errors='coerce'
)

print("Invalid Order Date:",
      df['Order Date'].isna().sum())

print("Invalid Ship Date:",
      df['Ship Date'].isna().sum())

# Kiểm tra giá trị nghiệp vụ
print("\nSales <= 0:", (df['Sales'] <= 0).sum())
print("Quantity <= 0:", (df['Quantity'] <= 0).sum())
print("Discount outside [0,1]:",
      (~df['Discount'].between(0, 1)).sum())

print("Order Date after Ship Date:",
      (df['Order Date'] > df['Ship Date']).sum())
```

```python
# Kiểm tra Customer ID và Customer Name
customer_check = (
    df.groupby('Customer ID')['Customer Name']
    .nunique()
)

print("Customer ID có nhiều Customer Name:",
      (customer_check > 1).sum())


# Kiểm tra Product ID và Product Name
product_check = (
    df.groupby('Product ID')['Product Name']
    .nunique()
)

print("Product ID có nhiều Product Name:",
      (product_check > 1).sum())


# Kiểm tra Order ID và Customer ID
order_check = (
    df.groupby('Order ID')['Customer ID']
    .nunique()
)

print("Order ID có nhiều Customer ID:",
      (order_check > 1).sum())
```

```python
product_mapping = (
    df.groupby('Product ID')['Product Name']
    .nunique()
)

invalid_products = product_mapping[
    product_mapping > 1
].index

product_errors = df[
    df['Product ID'].isin(invalid_products)
][['Product ID', 'Product Name']].drop_duplicates()

display(product_errors.sort_values('Product ID'))

print("Total inconsistent Product IDs:",
      len(invalid_products))
```

```python
print("Order Date range:")
print(df['Order Date'].min(), "to", df['Order Date'].max())

print("\nShip Date range:")
print(df['Ship Date'].min(), "to", df['Ship Date'].max())

print("\nCountry:")
print(df['Country'].unique())

print("\nPostal Code data type:")
print(df['Postal Code'].dtype)

print("\nNumber of unique Postal Codes:")
print(df['Postal Code'].nunique())
```

```python
# Thống kê số tên của từng Product ID
product_mapping = (
    df.groupby(['Product ID', 'Product Name'])
      .size()
      .reset_index(name='Frequency')
)

# Lấy những Product ID có nhiều tên
conflict_ids = (
    df.groupby('Product ID')['Product Name']
      .nunique()
)

conflict_ids = conflict_ids[
    conflict_ids > 1
].index

# Tạo báo cáo
product_quality_report = product_mapping[
    product_mapping['Product ID'].isin(conflict_ids)
].sort_values(['Product ID', 'Frequency'],
              ascending=[True, False])

print("Total conflicting Product IDs:",
      len(conflict_ids))

print("Total conflicting Product ID-Name pairs:",
      len(product_quality_report))

display(product_quality_report)

# Xuất báo cáo
product_quality_report.to_csv(
    'Product_Data_Quality_Report.csv',
    index=False,
    encoding='utf-8-sig'
)
```

```python
# Sao chép dữ liệu nguồn
df_stg = df.copy()

# Chuẩn hóa tên cột
df_stg.columns = (
    df_stg.columns
    .str.strip()
)

# Chuyển kiểu dữ liệu ngày tháng
df_stg['Order Date'] = pd.to_datetime(
    df_stg['Order Date'],
    errors='coerce'
)

df_stg['Ship Date'] = pd.to_datetime(
    df_stg['Ship Date'],
    errors='coerce'
)

# Postal Code là mã định danh, không phải số để tính toán
df_stg['Postal Code'] = (
    df_stg['Postal Code']
    .astype('string')
)

# Chuẩn hóa các trường văn bản
text_columns = [
    'Order ID',
    'Ship Mode',
    'Customer ID',
    'Customer Name',
    'Segment',
    'Country',
    'City',
    'State',
    'Postal Code',
    'Region',
    'Product ID',
    'Category',
    'Sub-Category',
    'Product Name'
]

for col in text_columns:
    df_stg[col] = df_stg[col].str.strip()

print("Staging shape:", df_stg.shape)
print("Missing values:", df_stg.isnull().sum().sum())

display(df_stg.head())
```

```python
Dim_Customer = (
    df_stg[
        ['Customer ID', 'Customer Name', 'Segment']
    ]
    .drop_duplicates()
    .sort_values('Customer ID')
    .reset_index(drop=True)
)

# Tạo Surrogate Key
Dim_Customer.insert(
    0,
    'CustomerKey',
    range(1, len(Dim_Customer) + 1)
)

# Đổi tên cột theo quy ước DW
Dim_Customer.rename(columns={
    'Customer ID': 'CustomerID',
    'Customer Name': 'CustomerName'
}, inplace=True)

print("Dim_Customer:", Dim_Customer.shape)

display(Dim_Customer.head())
```

```python
Dim_Product = (
    df_stg[
        [
            'Product ID',
            'Product Name',
            'Category',
            'Sub-Category'
        ]
    ]
    .drop_duplicates()
    .sort_values(
        ['Product ID', 'Product Name']
    )
    .reset_index(drop=True)
)

# Tạo Surrogate Key
Dim_Product.insert(
    0,
    'ProductKey',
    range(1, len(Dim_Product) + 1)
)

# Đổi tên cột
Dim_Product.rename(columns={
    'Product ID': 'ProductID',
    'Product Name': 'ProductName',
    'Sub-Category': 'SubCategory'
}, inplace=True)

print("Dim_Product:", Dim_Product.shape)

display(Dim_Product.head())
```

```python
Dim_Location = (
    df_stg[
        [
            'Country',
            'State',
            'City',
            'Postal Code',
            'Region'
        ]
    ]
    .drop_duplicates()
    .sort_values(
        ['Country', 'State', 'City', 'Postal Code']
    )
    .reset_index(drop=True)
)

Dim_Location.insert(
    0,
    'LocationKey',
    range(1, len(Dim_Location) + 1)
)

Dim_Location.rename(columns={
    'Postal Code': 'PostalCode'
}, inplace=True)

print("Dim_Location:", Dim_Location.shape)

display(Dim_Location.head())
```

```python
Dim_ShipMode = (
    df_stg[['Ship Mode']]
    .drop_duplicates()
    .sort_values('Ship Mode')
    .reset_index(drop=True)
)

Dim_ShipMode.insert(
    0,
    'ShipModeKey',
    range(1, len(Dim_ShipMode) + 1)
)

Dim_ShipMode.rename(columns={
    'Ship Mode': 'ShipMode'
}, inplace=True)

print("Dim_ShipMode:", Dim_ShipMode.shape)

display(Dim_ShipMode)
```

```python
# Xác định khoảng thời gian
start_date = min(
    df_stg['Order Date'].min(),
    df_stg['Ship Date'].min()
)

end_date = max(
    df_stg['Order Date'].max(),
    df_stg['Ship Date'].max()
)

# Tạo lịch liên tục
date_range = pd.date_range(
    start=start_date,
    end=end_date,
    freq='D'
)

Dim_Date = pd.DataFrame({
    'Date': date_range
})

# Tạo DateKey theo định dạng YYYYMMDD
Dim_Date['DateKey'] = (
    Dim_Date['Date'].dt.strftime('%Y%m%d')
    .astype('int64')
)

# Các thuộc tính thời gian
Dim_Date['Year'] = Dim_Date['Date'].dt.year

Dim_Date['Quarter'] = (
    Dim_Date['Date'].dt.quarter
)

Dim_Date['Month'] = (
    Dim_Date['Date'].dt.month
)

Dim_Date['MonthName'] = (
    Dim_Date['Date'].dt.month_name()
)

Dim_Date['Day'] = (
    Dim_Date['Date'].dt.day
)

Dim_Date['DayOfWeek'] = (
    Dim_Date['Date'].dt.dayofweek + 1
)

Dim_Date['DayName'] = (
    Dim_Date['Date'].dt.day_name()
)

Dim_Date['IsWeekend'] = (
    Dim_Date['Date'].dt.dayofweek >= 5
)

print("Dim_Date:", Dim_Date.shape)

display(Dim_Date.head())
```

```python
dimensions = {
    'Dim_Customer': Dim_Customer,
    'Dim_Product': Dim_Product,
    'Dim_Location': Dim_Location,
    'Dim_ShipMode': Dim_ShipMode,
    'Dim_Date': Dim_Date
}

for name, table in dimensions.items():

    print(f"\n===== {name} =====")

    print("Rows:", len(table))
    print("Columns:", len(table.columns))
    print("Duplicate rows:", table.duplicated().sum())
    print("Missing values:", table.isnull().sum().sum())
```

```python
Fact_Sales = df_stg[
    [
        'Row ID',
        'Order ID',
        'Customer ID',
        'Product ID',
        'Product Name',
        'Country',
        'State',
        'City',
        'Postal Code',
        'Region',
        'Ship Mode',
        'Order Date',
        'Ship Date',
        'Sales',
        'Quantity',
        'Discount',
        'Profit'
    ]
].copy()

# Đổi tên cột
Fact_Sales.rename(columns={
    'Row ID': 'RowID',
    'Order ID': 'OrderID',
    'Customer ID': 'CustomerID',
    'Product ID': 'ProductID',
    'Product Name': 'ProductName',
    'Postal Code': 'PostalCode',
    'Ship Mode': 'ShipMode',
    'Order Date': 'OrderDate',
    'Ship Date': 'ShipDate'
}, inplace=True)

print("Initial Fact shape:", Fact_Sales.shape)

display(Fact_Sales.head())
```

```python
# 1. Lookup CustomerKey
Fact_Sales = Fact_Sales.merge(
    Dim_Customer[
        ['CustomerKey', 'CustomerID']
    ],
    on='CustomerID',
    how='left',
    validate='many_to_one'
)

# 2. Lookup ProductKey
Fact_Sales = Fact_Sales.merge(
    Dim_Product[
        ['ProductKey', 'ProductID', 'ProductName']
    ],
    on=['ProductID', 'ProductName'],
    how='left',
    validate='many_to_one'
)

# 3. Lookup LocationKey
Fact_Sales = Fact_Sales.merge(
    Dim_Location[
        [
            'LocationKey',
            'Country',
            'State',
            'City',
            'PostalCode',
            'Region'
        ]
    ],
    on=[
        'Country',
        'State',
        'City',
        'PostalCode',
        'Region'
    ],
    how='left',
    validate='many_to_one'
)

# 4. Lookup ShipModeKey
Fact_Sales = Fact_Sales.merge(
    Dim_ShipMode[
        ['ShipModeKey', 'ShipMode']
    ],
    on='ShipMode',
    how='left',
    validate='many_to_one'
)

# 5. Lookup OrderDateKey
Dim_Date_Lookup = Dim_Date[
    ['Date', 'DateKey']
].copy()

Fact_Sales = Fact_Sales.merge(
    Dim_Date_Lookup.rename(columns={
        'Date': 'OrderDate',
        'DateKey': 'OrderDateKey'
    }),
    on='OrderDate',
    how='left',
    validate='many_to_one'
)

# 6. Lookup ShipDateKey
Fact_Sales = Fact_Sales.merge(
    Dim_Date_Lookup.rename(columns={
        'Date': 'ShipDate',
        'DateKey': 'ShipDateKey'
    }),
    on='ShipDate',
    how='left',
    validate='many_to_one'
)

print("Fact after lookup:", Fact_Sales.shape)

display(Fact_Sales.head())
```

```python
# Tạo khóa chính cho Fact
Fact_Sales.insert(
    0,
    'SalesKey',
    range(1, len(Fact_Sales) + 1)
)

# Tính thời gian giao hàng (ngày)
Fact_Sales['ShippingDays'] = (
    Fact_Sales['ShipDate'] -
    Fact_Sales['OrderDate']
).dt.days

# Chọn các cột cuối cùng
Fact_Sales = Fact_Sales[
    [
        'SalesKey',
        'RowID',
        'OrderID',
        'CustomerKey',
        'ProductKey',
        'LocationKey',
        'ShipModeKey',
        'OrderDateKey',
        'ShipDateKey',
        'Sales',
        'Quantity',
        'Discount',
        'Profit',
        'ShippingDays'
    ]
].copy()

print("Final Fact shape:", Fact_Sales.shape)

display(Fact_Sales.head())
```

```python
# Kiểm tra số lượng dòng
print("Fact rows:", len(Fact_Sales))

# Kiểm tra NULL
print("\nMissing values:")
print(Fact_Sales.isnull().sum())

# Kiểm tra trùng lặp
print("\nDuplicate rows:",
      Fact_Sales.duplicated().sum())

# Kiểm tra khóa chính
print("\nDuplicate SalesKey:",
      Fact_Sales['SalesKey'].duplicated().sum())

# Kiểm tra khóa ngoại
foreign_keys = {
    'CustomerKey': Dim_Customer['CustomerKey'],
    'ProductKey': Dim_Product['ProductKey'],
    'LocationKey': Dim_Location['LocationKey'],
    'ShipModeKey': Dim_ShipMode['ShipModeKey'],
    'OrderDateKey': Dim_Date['DateKey'],
    'ShipDateKey': Dim_Date['DateKey']
}

for fk, dimension_keys in foreign_keys.items():

    invalid = (
        ~Fact_Sales[fk].isin(dimension_keys)
    ).sum()

    print(f"{fk}: Invalid = {invalid}")
```

```python
import os
import zipfile
from google.colab import files

# 1. Danh sách các bảng cần xuất
tables = {
    'Dim_Customer': Dim_Customer,
    'Dim_Product': Dim_Product,
    'Dim_Location': Dim_Location,
    'Dim_ShipMode': Dim_ShipMode,
    'Dim_Date': Dim_Date,
    'Fact_Sales': Fact_Sales
}

# 2. Tạo thư mục chứa các file CSV
export_folder = '/content/DW_Export'
os.makedirs(export_folder, exist_ok=True)

# 3. Xuất từng bảng thành CSV
for table_name, table_df in tables.items():
    file_path = f'{export_folder}/{table_name}.csv'

    table_df.to_csv(
        file_path,
        index=False,
        encoding='utf-8-sig'
    )

    print(f'{table_name}: {len(table_df)} rows - Exported')

# 4. Nén tất cả thành một file ZIP
zip_path = '/content/DW_Export.zip'

with zipfile.ZipFile(zip_path, 'w') as zipf:
    for table_name in tables:
        file_path = f'{export_folder}/{table_name}.csv'
        zipf.write(file_path, arcname=f'{table_name}.csv')

print("\nExport completed!")

# 5. Tải ZIP về máy
files.download(zip_path)
```
